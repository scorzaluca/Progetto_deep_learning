# Retrieve Optuna Study

Load and inspect a completed Optuna study from the SQLite database.

```python
import optuna
```

## Load Study

Change `study_name` to the name of the study you want to retrieve.

```python
# Configuration
study_name = "lstm_48_1_1_expanding"  # Change this to your study name
storage_path = "sqlite:///../results/optuna_studies.db"

# Load study
study = optuna.load_study(study_name=study_name, storage=storage_path)
print(f"Study '{study_name}' loaded successfully.")
print(f"Number of trials: {len(study.trials)}")
```

## Best Trial

```python
print("Best MASE:", study.best_value)
print("\nBest Params:")
for key, value in study.best_params.items():
    print(f"  {key}: {value}")
```

## All Trials Summary

```python
# Convert to DataFrame for easy analysis
trials_df = study.trials_dataframe()
trials_df.head(10)
```

## List All Studies in Database

```python
# List all available studies
all_studies = optuna.study.get_all_study_names(storage=storage_path)
print("Available studies:")
for name in all_studies:
    print(f"  - {name}")
```
