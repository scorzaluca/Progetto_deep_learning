# Optuna Study Retrieval

This notebook provides utilities to load, inspect, and manage completed Optuna studies from the SQLite database.

```python
import optuna
```

## Load Study

Load a specific study from the database. Update the `study_name` variable to match the study you want to retrieve.

```python
# Configuration
study_name = "encoderlstm_24"
storage_path = "sqlite:///../results/optuna_studies.db"

# Load study from database
study = optuna.load_study(study_name=study_name, storage=storage_path)
print(f"Study '{study_name}' loaded successfully.")
print(f"Number of trials: {len(study.trials)}")
```

## Best Trial

Display the best trial's score and hyperparameters.

```python
print("Best MASE:", study.best_value)
print("\nBest Params:")
for key, value in study.best_params.items():
    print(f"  {key}: {value}")
```

## All Trials Summary

Convert all trials to a DataFrame for comprehensive analysis of the optimization history.

```python
# Convert trials to DataFrame for analysis
trials_df = study.trials_dataframe()
trials_df.head(10)
```

## List All Studies

Display all available studies stored in the database.

```python
# List all available studies in the database
storage_path = "sqlite:///../results/optuna_studies.db"
all_studies = optuna.study.get_all_study_names(storage=storage_path)
print("Available studies:")
for name in all_studies:
    print(f"  - {name}")
```

## Delete Studies

Remove obsolete or unwanted studies from the database. Update the `studies_to_delete` list as needed.

```python
# Delete multiple studies from the database
studies_to_delete = [
    "screening_patchtst_20251206_122521",
    "screening_patchtst_20251206_172308",
    "dlineari_prova",
    "dlineari_prova1",
    "dlineari_prova2",
    "dlineari_prova3",
    "dlinearm_prova1",
    "screening_lstm_20251206_122058",
    "screening_lstm_20251206_143140",
    "screening_lstm_20251206_172306"
]

for study_name in studies_to_delete:
    try:
        optuna.delete_study(study_name=study_name, storage=storage_path)
        print(f"Study '{study_name}' deleted successfully.")
    except KeyError:
        print(f"Error: Study '{study_name}' does not exist in the database.")
```
