# Compute Model Parameters

Questo notebook calcola `INPUT_SIZE` e `TARGET_IDX` dal DataLoader.
Questi valori devono essere copiati in `src/config/model_config.py`.

```python
import sys
import os

# Aggiungi project root al path per gli import
current_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(current_dir, '..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

print(f"Project Root: {project_root}")
```

```python
import pandas as pd
from src.DataLoading.data_loader import TS_Cross_Validator
from src.config import TARGET_COL, SAMPLING_CONFIG

# Carica il dataset preprocessato
DATA_PATH = "../data/processed/preprocessed_ds.csv"
df = pd.read_csv(DATA_PATH)

print(f"Dataset shape: {df.shape}")
print(f"Target column: {TARGET_COL}")
print(f"\nColonne ({len(df.columns)}):")
for i, col in enumerate(df.columns):
    print(f"  {i}: {col}")
```

```python
# Crea il DataLoader
validator = TS_Cross_Validator(df, TARGET_COL, SAMPLING_CONFIG)

# Prendi solo il primo fold per calcolare i parametri
train_loader, _, _ = next(validator.get_folds())

# Leggi i parametri dal dataset dentro il loader
dataset = train_loader.dataset
INPUT_SIZE = dataset.data_tensor.shape[1]
TARGET_IDX = dataset.target_col_idx

print("=" * 60)
print("PARAMETRI DA COPIARE IN model_config.py")
print("=" * 60)
print(f"\nINPUT_SIZE = {INPUT_SIZE}")
print(f"TARGET_IDX = {TARGET_IDX}")
print(f"\n# Verifica: INPUT_SIZE corrisponde a {INPUT_SIZE} feature")
print(f"# Verifica: TARGET_IDX={TARGET_IDX} corrisponde alla colonna '{df.columns[TARGET_IDX]}'")
```
